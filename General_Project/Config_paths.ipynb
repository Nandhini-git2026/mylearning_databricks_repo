
# This notebook is going do 2 activities

## 1. Once for all - Create all necessary path and table components needed for this project such as
## Catalog(parameter) -> Schema(parameter) -> Volume (sourcedatalake ,bronze ,silver ,gold).
## 2. Defining all configuration parameters such as volume path, table information etc., once for all, rather than we hardcode all these in every notebook

```python
# Widget creation

dbutils.widgets.text("catalog","prodcatalogdefault")
CATALOG=dbutils.widgets.get("catalog").strip()

dbutils.widgets.text("schema","prodschemadefault")
SCHEMA=dbutils.widgets.get("schema").strip()

print("CATALOG",CATALOG)
print("SCHEMA",SCHEMA)
```

```python
# Catalog/schema/volume creation

spark.sql(f"create catalog if not exists {CATALOG}")
spark.sql(f"create schema if not exists {CATALOG}.{SCHEMA};")


spark.sql(f"create volume if not exists {CATALOG}.{SCHEMA}.datalake;")
spark.sql(f"create volume if not exists {CATALOG}.{SCHEMA}.bronze;")
spark.sql(f"create volume if not exists {CATALOG}.{SCHEMA}.silver;")
spark.sql(f"create volume if not exists {CATALOG}.{SCHEMA}.gold;")
```

```python
# Assigning varaible to Catalog/schema/volume paths

SRC = f"Volume/{CATALOG}/{SCHEMA}/datalake"
BRONZE = f"Volume/{CATALOG}/{SCHEMA}/bronze"
SILVER = f"Volume/{CATALOG}/{SCHEMA}/silver"
GOLD = f"Volume/{CATALOG}/{SCHEMA}/gold"

SILVERDB = f"{CATALOG}.{SCHEMA}"
GOLDB = f"{CATALOG}.{SCHEMA}"

print("Source datalake location",SRC)
print("BRONZE datalake location",BRONZE)
print("SILVERDB",SILVERDB)
print("GOLDB",GOLDB)

```

```python
# exit notebook with path as string
# dumps dictionary to string

import json

dbutils.notebook.exit(
json.dumps({
    "CATALOG":CATALOG,
    "SCHEMA":SCHEMA,
    "SRC":SRC,
    "BRONZE":BRONZE,
    "GOLD":GOLD,
    "SILVERDB":SILVERDB,
    "GOLDB":GOLDB,
})
)
```
