```python
customer_df = spark.read.csv("/Volumes/we48/izdata/data/c_customer.csv",sep="|",inferSchema=True,header=True)
trans_df = spark.read.csv("/Volumes/we48/izdata/data/c_txns.csv",sep="|",inferSchema=True,header=True)

customer_df.show()
trans_df.show()

```

```python
# inner join

trans_cust_inner_join_df = trans_df.join(customer_df,on ="cid",how="inner") #or
trans_cust_inner_join_df = trans_df.join(customer_df,customer_df.cid==trans_df.cid,how="inner")
trans_cust_inner_join_df.show()
```

```python
# left join - All left rows + matching right rows
trans_cust_left_join_df = trans_df.join(customer_df,on ="cid",how="left") #or
trans_cust_left_join_df = trans_df.join(customer_df,customer_df.cid==trans_df.cid,how="left")

trans_cust_left_join_df.show()
```

```python
# right join - All right rows + matching left rows

trans_cust_right_join_df = trans_df.join(customer_df,on ="cid",how="right")

trans_cust_right_join_df.show()
```

```python
# outer join - All rows from both DataFrames

trans_cust_outer_join_df = trans_df.join(customer_df,on ="cid",how="outer") # outer or full or full_outer

trans_cust_outer_join_df.show()
```

```python
# left semi - keep left rows that have a match, but return only the left table's columns. Also runs faster in Spark since no data from the right side is carried.

trans_cust_left_semi_join_df = trans_df.join(customer_df,on ="cid",how="left_semi")

trans_cust_left_semi_join_df.show()
```

```python
# left anti - returns only the rows from the left DataFrame that do not have a matching row in the right DataFrame.

trans_cust_left_anti_join_df = trans_df.join(customer_df,on ="cid",how="left_anti")

trans_cust_left_anti_join_df.show()
```

```python
# Cross join
emp_data =[(100,"a1",23),(101,"a2",25),(102,"a3",27),(103,"a4",29),(104,"a5",31),(105,"a6",33),(105,"a7",35),(None,"jeeva",21)]

city_data =[(100,"Chennai"),(101,"Bangalore"),(102,"Mumbai"),(103,"Delhi"),(None,"NJ")]

emp_df = spark.createDataFrame(emp_data,["eid","ename","age"])
city_df = spark.createDataFrame(city_data,["eid","city"])

emp_df.join(city_df,on="eid",how="cross").show()
```

```python
# eqNullSafe - is used to compare two columns while treating NULL values as equal. It is the equivalent of SQL's <=> (null-safe equality) operator.

# we will not use regularly , on need basis 

emp_data =[(100,"a1",23),(101,"a2",25),(102,"a3",27),(103,"a4",29),(104,"a5",31),(105,"a6",33),(105,"a7",35),(None,"jeeva",21)]

city_data =[(100,"Chennai"),(101,"Bangalore"),(102,"Mumbai"),(103,"Delhi"),(None,"NJ")]

emp_df = spark.createDataFrame(emp_data,["eid","ename","age"])
city_df = spark.createDataFrame(city_data,["eid","city"])

emp_df.join(city_df,emp_df.eid.eqNullSafe(city_df.eid),how="inner").show()

```
