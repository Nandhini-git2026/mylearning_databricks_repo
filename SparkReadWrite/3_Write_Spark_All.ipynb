# write dataframe into parquet/orc/csv/xml/json/table

```python
schema_str = ("custid int, fname string,lname string,age int,profession string")
df = spark.read.csv("/Volumes/we48/izdata/data", header=True, schema=schema_str)
# csv save
df.write.mode("overwrite")\
  .format("csv")\
  .save("/Volumes/we48/izdata/data_custs")

# json save
df.write.mode("overwrite")\
  .format("json")\
  .option("compression","gzip")\
  .save("/Volumes/we48/izdata/data_custs/json")

# xml save
df.write.mode("overwrite")\
  .format("xml")\
  .option("rootTag","Customer")\
  .option("rowTag","Person")\
  .save("/Volumes/we48/izdata/data_custs/xml")

#parquet
df.write.mode("overwrite")\
  .format("parquet")\
  .save("/Volumes/we48/izdata/data_custs/parquet")

#orc
df.write.mode("overwrite")\
.format("orc")\
.save("/Volumes/we48/izdata/data_custs/orc")

#save in table
df.write.mode("overwrite").saveAsTable("izwe48catalog.we48db.tbl_customer")
spark.sql("select count(1)  from izwe48catalog.we48db.tbl_customer").show()

#insert into, this will append
df.write.mode("append").insertInto("izwe48catalog.we48db.tbl_customer")
spark.sql("select count(1)  from izwe48catalog.we48db.tbl_customer").show()


```
