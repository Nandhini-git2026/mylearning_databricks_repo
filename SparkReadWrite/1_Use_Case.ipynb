```python
# /Volumes/izwe48catalog/we48db/staging/usecase/  Files available here

from pyspark.sql.functions import lit,concat,col,current_date,split,regexp_extract,current_user

# Creating df each per region

emea_df1 = spark.read.csv("/Volumes/izwe48catalog/we48db/staging/usecase/empdata_EMEA_20260620*",header=True,sep=",")
nam_df1 = spark.read.csv("/Volumes/izwe48catalog/we48db/staging/usecase/empdata_NAM_20260620*",header=True,sep=",")
apac_df1 = spark.read.csv("/Volumes/izwe48catalog/we48db/staging/usecase/empdata_APAC_20260620*",header=True,sep=",")

# filename from path/ split file name for source and source date

source_df1 = emea_df1.withColumn("file_name", col("_metadata.file_name")).withColumn("Source", split("file_name", "_")[1]).withColumn("data dt",regexp_extract( col("_metadata.file_name"),r'(\d{8})', 1))
source_df2 = nam_df1.withColumn("file_name", col("_metadata.file_name")).withColumn("Source", split("file_name", "_")[1]).withColumn("data dt", regexp_extract( col("_metadata.file_name"),r'(\d{8})', 1))
source_df3 = apac_df1.withColumn("file_name", col("_metadata.file_name")).withColumn("Source", split("file_name", "_")[1]).withColumn("data dt", regexp_extract( col("_metadata.file_name"), r'(\d{8})', 1))

#combining all 3 dfs
combined_df = source_df1.union(source_df2).union(source_df3)

#final select
final_df = combined_df.select("cid",concat(col("fname"),lit(" "),col("lname")).alias("fullname"),"age",col("prof").alias("Profession"),"source","data dt").withColumn("load date",current_date()).withColumn("Created by",current_user())
#display(final_df)
final_df.show(50,True)


```
